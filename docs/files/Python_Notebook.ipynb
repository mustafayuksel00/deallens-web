```python
import importlib.metadata
import joblib
import numpy as np
import pandas as pd
from catboost import CatBoostClassifier
```

```python
from sklearn.base import clone
from sklearn.dummy import DummyClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import StratifiedGroupKFold, cross_validate
from sklearn.metrics import (accuracy_score, precision_score, recall_score, f1_score,
    average_precision_score, roc_auc_score, confusion_matrix, precision_recall_curve)
```

```python
VERSION = 'deallens-python-v2-2026-10-08'
SEED = 8102026
FEATURE_MAP = {'Attr2': 'liabilities_to_assets', 'Attr4': 'current_ratio',
    'Attr9': 'sales_to_assets', 'Attr21': 'sales_growth_ratio',
    'Attr42': 'operating_profit_to_sales'}
FEATURES = list(FEATURE_MAP.values())
FEATURES64 = [f'Attr{i}' for i in range(1, 65)]
```

```python
TARGET = 'bankruptcy_label'
LIMITATION = ('Bu veri önceki deneylerde incelendi. Bu sonuçlar yeniden üretilebilir '
    'eğitim karşılaştırmasıdır; yeni bağımsız dış test veya güncel şirket başarısı değildir.')

def frame_groups(df):
    # Target and record_id MUST NOT determine grouping. Identical inputs stay together.
    return pd.util.hash_pandas_object(df[FEATURES64], index=False).astype(str)

def partitions(df):
    groups = frame_groups(df)
    required = {'partition', 'partition_order', 'cv_fold'}
    assert required.issubset(df.columns), 'CSV sabit ayrım sütunlarını içermelidir.'
    assert set(df['partition']) == {'train', 'validation', 'test'}
    assert df.loc[df['partition']!='train', 'cv_fold'].eq(-1).all()
    assert set(df.loc[df['partition']=='train', 'cv_fold']) == {0,1,2}
    split = {p: df[df['partition']==p].sort_values('partition_order').index.to_numpy()
             for p in ['test','validation','train']}
    for p,rows in split.items():
        assert df.iloc[rows]['partition_order'].tolist() == list(range(len(rows)))
    for a, b in [('train', 'test'), ('train', 'validation'), ('test', 'validation')]:
        assert not set(groups.iloc[split[a]]) & set(groups.iloc[split[b]])
    assert len(set(np.concatenate(list(split.values())))) == len(df)
    return split, groups

def cat_model(depth=4, indicator=False):
    return Pipeline([('imputer', SimpleImputer(strategy='median', add_indicator=indicator)),
        ('model', CatBoostClassifier(iterations=300, depth=depth, learning_rate=0.035,
            l2_leaf_reg=3, random_strength=1, border_count=64, loss_function='Logloss',
            random_seed=SEED, thread_count=4, verbose=False, allow_writing_files=False))])
```

```python
def candidates():
    return {
        'Dummy 5': (FEATURES, Pipeline([('imputer', SimpleImputer(strategy='median')),
            ('model', DummyClassifier(strategy='most_frequent'))])),
        'Decision Tree 5': (FEATURES, Pipeline([('imputer', SimpleImputer(strategy='median')),
            ('model', DecisionTreeClassifier(max_depth=4, min_samples_leaf=20,
                class_weight='balanced', random_state=SEED))])),
        'Logistic Regression 5': (FEATURES, Pipeline([('imputer', SimpleImputer(strategy='median')),
            ('scale', StandardScaler()), ('model', LogisticRegression(max_iter=3000,
                class_weight='balanced', random_state=SEED))])),
        'Random Forest 5': (FEATURES, Pipeline([('imputer', SimpleImputer(strategy='median')),
            ('model', RandomForestClassifier(n_estimators=200, max_depth=8,
                min_samples_leaf=10, class_weight='balanced', random_state=SEED, n_jobs=4))])),
        'CatBoost 5 depth4': (FEATURES, cat_model()),
        'CatBoost 64 depth4': (FEATURES64, cat_model()),
        'CatBoost 64 depth5': (FEATURES64, cat_model(depth=5)),
        'CatBoost 64 indicators depth4': (FEATURES64, cat_model(indicator=True)),
        'CatBoost 64 indicators depth5': (FEATURES64, cat_model(depth=5, indicator=True)),
    }

def threshold_for_f1(y, score):
    p, r, t = precision_recall_curve(y, score)
    f = 2*p[:-1]*r[:-1] / np.maximum(p[:-1]+r[:-1], 1e-12)
    return float(t[np.argmax(f)])

def metrics(y, score, threshold):
    pred = (score >= threshold).astype(int)
    return {'threshold': float(threshold), 'accuracy': float(accuracy_score(y, pred)),
        'precision': float(precision_score(y, pred, zero_division=0)),
        'recall': float(recall_score(y, pred, zero_division=0)),
        'f1': float(f1_score(y, pred, zero_division=0)),
        'average_precision': float(average_precision_score(y, score)),
        'roc_auc': float(roc_auc_score(y, score)),
        'confusion_matrix': confusion_matrix(y, pred, labels=[0, 1]).tolist()}

def validate_input(frame, features):
    if not isinstance(frame, pd.DataFrame) or frame.empty:
        raise ValueError('En az bir kayıt içeren tablo gerekli.')
    if frame.columns.duplicated().any():
        raise ValueError('Yinelenen sütun adı var.')
    missing = set(features) - set(frame.columns)
    if missing:
        raise ValueError(f'Eksik oran sütunları: {sorted(missing)}')
    numeric = frame[features].apply(pd.to_numeric, errors='raise')
    if np.isinf(numeric.to_numpy(dtype=float)).any():
        raise ValueError('Sonsuz değer kabul edilmez.')
    if numeric.isna().all(axis=1).any():
        raise ValueError('Bütün oranları eksik kayıt kabul edilmez.')
    return numeric

def predict_bundle(bundle, frame):
    numeric = validate_input(frame, bundle['features'])
    scores = bundle['model'].predict_proba(numeric)[:, 1]
    return pd.DataFrame({'model_score': scores,
        'predicted_label': (scores >= bundle['threshold']).astype(int),
        'missing_features': numeric.isna().sum(axis=1).to_numpy(),
        'model_version': bundle['version']})
```

```python
def run_experiment(df64, df5, output_dir):
    output_dir = Path(output_dir)
    output_dir.mkdir(parents=True, exist_ok=True)
    assert df64['record_id'].is_unique and df5['record_id'].is_unique
    assert df64['record_id'].tolist() == df5['record_id'].tolist()
    assert df64[TARGET].equals(df5[TARGET])
    for column in ['partition','partition_order','cv_fold']:
        assert df64[column].equals(df5[column]), 'İki CSV aynı ayrımı taşımalıdır.'
    np.testing.assert_allclose(df64[list(FEATURE_MAP)].to_numpy(),
        df5[FEATURES].to_numpy(), equal_nan=True)
    validate_input(df64, FEATURES64)
    assert df64[TARGET].notna().all() and set(df64[TARGET].unique()) == {0, 1}
    df = df64.copy()
    for src, dst in FEATURE_MAP.items():
        df[dst] = df[src]
    split, groups = partitions(df)
    manifest = df[['record_id', TARGET]].copy()
    manifest['group_id'] = groups
    for part, rows in split.items():
        manifest.loc[rows, 'partition'] = part
    manifest.to_csv(output_dir/'split_manifest.csv', index=False)
    configs = candidates()
    protocol = {'version': VERSION, 'seed': SEED, 'grouping': 'all 64 inputs, excluding target',
        'split': 'Precomputed group-stratified 5-fold partition and 3-fold training CV in CSV metadata; stable across library versions',
        'selection': 'highest mean train-only 3-fold group CV average_precision among 64-input candidates',
        'threshold': 'max F1 on separate validation partition; no refit afterwards',
        'external_validation': False, 'limitation': LIMITATION,
        'candidate_parameters': {n: m.named_steps['model'].get_params()
            for n, (_, m) in configs.items()},
        'source_csv_sha256': hashlib.sha256(df64.to_csv(index=False).encode()).hexdigest()}
    (output_dir/'protocol.json').write_text(json.dumps(protocol, ensure_ascii=False, indent=2), encoding='utf-8')
    train, val, test = split['train'], split['validation'], split['test']
    fold_ids = df.iloc[train]['cv_fold'].to_numpy()
    cv = [(np.flatnonzero(fold_ids!=i), np.flatnonzero(fold_ids==i)) for i in [0,1,2]]
    for fitting,heldout in cv:
        assert not set(groups.iloc[train[fitting]]) & set(groups.iloc[train[heldout]])
    cv_rows = []
    fitted = {}
    # No test predictions or scores during candidate fitting and selection.
    for name, (features, model) in configs.items():
        scores = cross_validate(clone(model), df.iloc[train][features], df.iloc[train][TARGET],
            groups=groups.iloc[train], cv=cv, scoring='average_precision', n_jobs=1,
            error_score='raise')['test_score']
        cv_rows.append({'model': name, 'feature_count': len(features),
            'cv_average_precision': float(scores.mean()), 'cv_std': float(scores.std()),
            'fold_average_precision': scores.tolist()})
        model.fit(df.iloc[train][features], df.iloc[train][TARGET])
        validation_score = model.predict_proba(df.iloc[val][features])[:, 1]
        threshold = 0.5 if name == 'Dummy 5' else threshold_for_f1(df.iloc[val][TARGET], validation_score)
        fitted[name] = {'model': model, 'threshold': threshold, 'features': features,
            'version': VERSION, 'name': name, 'training_record_ids': df.iloc[train]['record_id'].tolist(),
            'validation': metrics(df.iloc[val][TARGET], validation_score, threshold),
            'warning': LIMITATION}
        print(f'{name}: train CV AP={scores.mean():.4f}', flush=True)
    advanced = [r for r in cv_rows if r['feature_count'] == 64]
    selected = max(advanced, key=lambda r: r['cv_average_precision'])['model']
    selection = {'selected_model': selected, 'reason': protocol['selection'],
        'threshold': fitted[selected]['threshold'], 'cv': cv_rows}
    # Persist decision before revealing this rerun's test scores.
    (output_dir/'selection_before_test.json').write_text(json.dumps(selection, ensure_ascii=False, indent=2), encoding='utf-8')
    model_results = {}
    predictions = []
    for name, bundle in fitted.items():
        score = bundle['model'].predict_proba(df.iloc[test][bundle['features']])[:, 1]
        model_results[name] = {'validation': bundle['validation'],
            'train': metrics(df.iloc[train][TARGET], bundle['model'].predict_proba(df.iloc[train][bundle['features']])[:,1], bundle['threshold']),
            'test': metrics(df.iloc[test][TARGET], score, bundle['threshold'])}
        predictions.append(pd.DataFrame({'record_id': df.iloc[test]['record_id'].to_numpy(),
            'actual_label': df.iloc[test][TARGET].to_numpy(), 'score': score,
            'predicted_label': (score >= bundle['threshold']).astype(int), 'model': name}))
    pd.concat(predictions).to_csv(output_dir/'test_predictions.csv', index=False)
    joblib.dump(fitted[selected], output_dir/'deallens_catboost_deployment.joblib')
    joblib.dump(fitted['Decision Tree 5'], output_dir/'deallens_tree_5.joblib')
    # Group CV statistics are for selection, not unbiased final performance.
    result = {'version': VERSION, 'limitation': LIMITATION, 'selected_model': selected,
        'data': {'source_rows': len(df), 'rows_removed': 0, 'feature_counts': [5,64],
            'input_duplicate_rows': int(df[FEATURES64].duplicated().sum()),
            'class_counts': {str(k):int(v) for k,v in df[TARGET].value_counts().sort_index().items()},
            'partitions': {p:{'rows':len(i),'positive':int(df.iloc[i][TARGET].sum())} for p,i in split.items()},
            'missing_5': int(df[FEATURES].isna().sum().sum()),
            'missing_64': int(df[FEATURES64].isna().sum().sum())},
        'cv': cv_rows, 'models': model_results,
        'deployment': {'same_as_evaluated_model': True, 'refit_on_full_data': False,
            'threshold': fitted[selected]['threshold'], 'features': fitted[selected]['features']},
        'environment': {'python': platform.python_version(), 'packages': {
            p: importlib.metadata.version(p) for p in ['numpy','pandas','scipy','scikit-learn','catboost','joblib']}}}
    (output_dir/'results.json').write_text(json.dumps(result, ensure_ascii=False, indent=2), encoding='utf-8')
    return result, fitted, split
```

```python
import base64
import zlib
EMBEDDED_CSV = {'deallens_financial_risk.csv': {'sha256': '956418f9335489b0c54a37be9ee7f1f9ac16b39da8e85373f3c1e3873efee4a2', 'payload': ''}, 'deallens_financial_risk_64.csv': {'sha256': '2259c9ad2acc4644b1ff9a9775b9981b353d9e57707ed0b4eeedbe2b0a17c900', 'payload': ''}}
import matplotlib.pyplot as plt
from sklearn.metrics import ConfusionMatrixDisplay
from sklearn.tree import plot_tree
from IPython.display import display

def locate_csv(name):
    for path in [Path(name), Path('data')/name, Path('/content')/name]:
        if path.exists():
            return path
    item = EMBEDDED_CSV[name]
    raw = zlib.decompress(base64.b64decode(item['payload']))
    if hashlib.sha256(raw).hexdigest() != item['sha256']:
        raise ValueError('Gömülü CSV bütünlük kontrolü başarısız: '+name)
    path = Path(name)
    path.write_bytes(raw)
    print('Açık lisanslı CSV kopyası SHA256 kontrolüyle hazırlandı:',name)
    return path

df5 = pd.read_csv(locate_csv('deallens_financial_risk.csv'))
df64 = pd.read_csv(locate_csv('deallens_financial_risk_64.csv'))
print('5 oran:',df5.shape,'64 oran:',df64.shape)
display(df5.head())
```

## Test metrikleri
Precision: riskli dediğimiz kayıtların ne kadarı pozitif?
Recall: pozitif kayıtların ne kadarını yakaladık?
F1: precision ve recall dengesi. Average precision sıralama başarısını özetler,
burada trapezle hesaplanan PR eğrisi alanı olarak adlandırılmaz.
Eşik ve CV puanları seçim için kullanılır, tarafsız nihai başarı kanıtı sayılmaz.

```python
def tahmin_et(yukumluluk_varlik,cari_oran,satis_varlik,satis_buyume,faaliyet_marji):
    values = np.asarray([yukumluluk_varlik,cari_oran,satis_varlik,satis_buyume,faaliyet_marji],dtype=float)
    if not np.isfinite(values).all():
        raise ValueError('Beş girdinin tamamı sonlu sayı olmalıdır.')
    return predict_bundle(bundles['Decision Tree 5'],pd.DataFrame([values],columns=FEATURES))

for name,values in [('Dengeli profil',[.35,2,1.5,1.1,.1]),
    ('Finansal baskı profili',[.85,.7,.6,.85,-.15]),
    ('Büyüyen karma profil',[.6,1.2,2,1.25,.03])]:
    print(name)
    display(tahmin_et(*values))

print('64 oranlı tarihsel test örneği:')
display(predict_bundle(selected_bundle,df64.iloc[[split['test'][0]]]))

"""Notebook'a gömülen kullanıcı giriş akışı; model eğitimi içermez."""
import math


def kullanici_tahmini(tahmin_fonksiyonu, input_fn=None, output_fn=print):
    """Beş kaynak oranını al; hatalı girdiyi yeniden sor; q ile iptal et."""
    if input_fn is None:
        input_fn = input
    alanlar = [
        ('Yükümlülük / varlık', 'örnek: 0.35'),
        ('Cari oran', 'örnek: 2.0'),
        ('Satış / varlık', 'örnek: 1.5'),
        ('Cari / önceki dönem satış', '1.10 yaklaşık %10 artış demektir'),
        ('Faaliyet kârı / satış', 'örnek: 0.10'),
    ]
    output_fn('Beş oranlı karar ağacı ile eğitim amaçlı tahmin.')
    output_fn('Yüzde yerine kaynak oranını yazın: %35 için 0.35. Ondalık virgül kabul edilir. İptal: q.')
    values = []
    for label, explanation in alanlar:
        while True:
            try:
                raw = input_fn(f'{label} ({explanation}): ').strip()
            except (EOFError, KeyboardInterrupt):
                output_fn('Giriş iptal edildi; tahmin yapılmadı.')
                return None
            if raw.lower() in ('q', 'iptal'):
                output_fn('Giriş iptal edildi; tahmin yapılmadı.')
                return None
            try:
                value = float(raw.replace(',', '.'))
                if not math.isfinite(value):
                    raise ValueError
            except ValueError:
                output_fn('Geçerli, sonlu bir sayı yazın. Boş değer, metin, NaN ve sonsuz kabul edilmez.')
                continue
            values.append(value)
            break
    result = tahmin_fonksiyonu(*values)
    row = result.iloc[0]
    label = 'iflas etiketi / riskli sınıf' if int(row['predicted_label']) == 1 else 'iflas yok etiketi / risksiz sınıf'
    output_fn(f'Girdiğiniz beş orana göre model {label} tahmin etti; model skoru {float(row["model_score"]):.4f}.')
    output_fn('Bu skor gerçek iflas yüzdesi değildir. Risksiz sınıf tahmini güvence değildir; tarihsel eğitim modeli satın alma önerisi vermez.')
    return result


# Colab'da beş oranı girin. Girişsiz otomatik kontrolde diğer sonuçlar korunur.
from IPython.core.error import StdinNotImplementedError
try:
    kullanici_sonucu = kullanici_tahmini(tahmin_et)
except StdinNotImplementedError:
    print('Bu otomatik çalıştırma kullanıcı girişini desteklemiyor. Colab’da bu hücreyi elle çalıştırıp beş oranı girin.')
```
